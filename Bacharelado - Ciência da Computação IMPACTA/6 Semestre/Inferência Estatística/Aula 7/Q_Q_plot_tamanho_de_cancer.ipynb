```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

tamanho_cancer = [120, 43, 128, 50, 324, 656, 128, 436, 654, 347, 554]

# Gerando o Q-Q Plot
plt.figure(figsize=(6, 4))
stats.probplot(tamanho_cancer, dist="norm", plot=plt)

plt.title("Q-Q Plot: Tamanho do Câncer")
plt.xlabel("Quartis Teóricos (Normal)")
plt.ylabel("Quartis dos Dados (Tamanho)")
plt.grid(True)
plt.legend(['Dados dos Tumores','Reta Bissetriz (Normal)'], loc='upper left')
plt.show()
```

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt

hospital_A = [120, 43, 128, 50, 324, 656, 128, 436, 654, 347, 554]
hospital_B = [130, 45, 135, 55, 330, 660, 130, 440, 660, 350, 560]

# Comparando as duas amostras diretamente
pp_x = sm.ProbPlot(np.array(hospital_A))
pp_y = sm.ProbPlot(np.array(hospital_B))

fig = pp_x.qqplot(line='45', other=pp_y)
plt.title("Q-Q Plot: Comparando Hospital A vs Hospital B")
plt.show()
```
