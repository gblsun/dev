# Exercício MediaPipe — Computação Gráfica e RA/RV

Neste exercício, vamos criar uma aplicação de Realidade Aumentada com webcam usando MediaPipe + OpenCV.

Objetivos:
- detectar mãos em tempo real
- localizar landmarks em 2D
- sobrepor elementos gráficos na cena (AR)
- estudar interação entre visão computacional e gráficos em tempo real

```python
import cv2
import mediapipe as mp
import matplotlib.pyplot as plt
from IPython.display import display

print("Versão do MediaPipe:", getattr(mp, "__version__", "desconhecida"))
print("Tem solutions?", hasattr(mp, "solutions"))

if not hasattr(mp, "solutions"):
    raise ImportError(
        "MediaPipe instalado incorretamente.\n"
        "1) execute a célula de instalação acima;\n"
        "2) reinicie o kernel;\n"
        "3) execute esta célula novamente."
    )

# Inicialização do MediaPipe
mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils
mp_drawing_styles = mp.solutions.drawing_styles

# Configuração da webcam
cap = cv2.VideoCapture(0)
if not cap.isOpened():
    raise RuntimeError("Não foi possível abrir a webcam. Verifique a câmera do computador.")

hands = mp_hands.Hands(
    static_image_mode=False,
    max_num_hands=2,
    min_detection_confidence=0.7,
    min_tracking_confidence=0.5,
)

ret, frame = cap.read()
if not ret:
    raise RuntimeError("Não foi possível capturar um frame da webcam.")

frame = cv2.flip(frame, 1)
frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
resultados = hands.process(frame_rgb)

# Overlay de realidade aumentada
if resultados.multi_hand_landmarks:
    for hand_landmarks in resultados.multi_hand_landmarks:
        mp_drawing.draw_landmarks(
            frame,
            hand_landmarks,
            mp_hands.HAND_CONNECTIONS,
            mp_drawing_styles.get_default_hand_landmarks_style(),
            mp_drawing_styles.get_default_hand_connections_style(),
        )

        idx = 9
        x = int(hand_landmarks.landmark[idx].x * frame.shape[1])
        y = int(hand_landmarks.landmark[idx].y * frame.shape[0])

        cv2.circle(frame, (x, y), 35, (0, 255, 255), 2)
        cv2.circle(frame, (x, y), 18, (255, 0, 255), -1)

        tamanho = 90
        cv2.rectangle(frame, (x - tamanho, y - tamanho), (x + tamanho, y + tamanho), (0, 255, 0), 2)
        for dx, dy in [(-tamanho, -tamanho), (tamanho, -tamanho), (-tamanho, tamanho), (tamanho, tamanho)]:
            cv2.line(frame, (x + dx, y + dy), (x, y), (255, 255, 0), 2)

cv2.putText(
    frame,
    "AR com MediaPipe - Mão detectada",
    (20, 30),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.8,
    (255, 255, 255),
    2,
)

cap.release()

# Exibe a imagem dentro do notebook
frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
fig, ax = plt.subplots(figsize=(10, 7))
ax.imshow(frame_rgb)
ax.axis("off")
plt.tight_layout()
display(fig)
plt.close(fig)

print("Frame processado e exibido no notebook.")
```

## Explicação do exercício

Este exemplo mostra como a visão computacional pode ser integrada a aplicações de RA/RV:

1. A webcam captura imagens em tempo real.
2. O MediaPipe detecta pontos (landmarks) da mão.
3. Esses pontos são convertidos em coordenadas da imagem.
4. Elementos gráficos, como círculos e retângulos, são desenhados sobre a mão.
5. O efeito visual simula uma sobreposição de objetos virtuais no mundo real.
